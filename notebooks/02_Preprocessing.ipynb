# Fake News Detection System

## Data Preprocessing

### Objective

Prepare the LIAR dataset for model training.

### Tasks

- Load train, validation and test datasets
- Merge datasets
- Convert labels to binary classes
- Handle missing values
- Remove duplicate statements
- Save processed dataset

```python
import pandas as pd
import numpy as np
```

```python
fake_df = pd.read_csv("../data/raw/Fake.csv")
true_df = pd.read_csv("../data/raw/True.csv")
```

```python
fake_df["label"] = 0
true_df["label"] = 1
```

```python
df = pd.concat([fake_df, true_df],ignore_index=True)
```

```python
df["content"] = (
    df["title"] + " " + df["text"]
)
```

```python
df = df[
    [
        "content",
        "label"
    ]
]
```

```python
df.isnull().sum()
```

```python
df = df.dropna(
    subset=["content"]
).reset_index(drop=True)
```

```python
df = df.drop_duplicates(
    subset=["content"]
).reset_index(drop=True)
```

```python
print(df.shape)
print(df["label"].value_counts())
```

```python
df.to_csv(
    "../data/processed/preprocessed_news.csv",
    index=False
)
```
